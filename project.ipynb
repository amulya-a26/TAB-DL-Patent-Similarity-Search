```python
df['abstract_clean']=(df['abstract'].fillna('')
                      .astype(str)
                      .str.lower()
                      .str.replace(r'\s+',' ',regex=True)
                      .str.replace(r'[^a-z0-9\-\./\+\(\)\s]', ' ', regex=True)
                    .str.strip()          
                      )

df['title_clean'] = df['title'].fillna('').astype(str).str.lower().str.replace(r'\s+', ' ', regex=True).str.replace(r'[^a-z0-9\-\./\+\(\)\s]', ' ', regex=True).str.strip()
```

```python
df.head()
```

```python
print(df['abstract'].iloc[0])
print(df['abstract_clean'].iloc[0])
```

```python
df['text'] = df['title_clean'] + " " + df['abstract_clean']
df.head()
```

```python
# install sentence-transformers if missing
%pip install -q sentence-transformers

from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
text = df['text'].iloc[0]

embedding = model.encode(text)

print(embedding.shape)
```

```python
sample_texts = df['text'].head(100).tolist()

embeddings = model.encode(
    sample_texts,
    show_progress_bar=True
)
print(embeddings.shape)
```

```python
all_texts = df['text'].tolist()

embeddings = model.encode(
    all_texts,
    batch_size=32,
    show_progress_bar=True
)
print(embeddings.shape)
```

```python
np.save("patent_embeddings.npy", embeddings)
```

```python
patent_embeddings = pd.DataFrame(embeddings)
patent_embeddings.insert(0, "patent_id", df["patent_id"].values)

patent_embeddings.to_csv("patent_embeddings.csv", index=False)
np.save("patent_embeddings.npy", embeddings)

print("Saved patent_embeddings.npy and patent_embeddings.csv")
```
